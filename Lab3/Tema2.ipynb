Ex. 1 Se aruncă două zaruri, iar rezultatele sunt notate cu X şi Y.

1. Calculaţi probabilitatea ca suma să fie cel puţin 10, ştiind că primul zar indică un număr par.

Spatiul total de evenimente are $6 \times 6 = 36$ variante.

 P(suma ≥ 10 | X par)

 X par inseamna 18 variante posibile.

 Cazuri favorabile (X par si suma ≥ 10): (4,6), (6,4), (6,5), (6,6) -> 4 variante.

 Probabilitate = 4 / 18 = 2 / 9 (aprox. 0.2222)


2. Calculaţi probabilitatea ca suma să fie cel puţin 10, ştiind că cel puţin unul dintre zaruri indică un număr par.

P(suma ≥ 10 | cel putin un zar par)

Cel putin un zar par = Total (36) - Ambele impare (3x3=9) = 27 variante.

Cazuri favorabile (suma ≥ 10 cu cel putin un zar par): (4,6), (6,4), (5,6), (6,5), (6,6) -> 5 variante.

Probabilitate = 5 / 27 (aprox. 0.1852)

3. Calculaţi probabilitatea ca ambele zaruri să indice numere pare, ştiind că suma lor este 8.

P(ambele pare | suma = 8)

Suma 8: (2,6), (3,5), (4,4), (5,3), (6,2) -> 5 variante.

Cazuri favorabile (ambele pare): (2,6), (4,4), (6,2) -> 3 variante.

Probabilitate = 3 / 5 = 0.6

4. Sunt evenimentele A = {X + Y ≥ 10} şi B = {X este par} independente?

Independenta evenimentelor A = {X + Y ≥ 10} si B = {X este par}

P(A) = 6 / 36 = 1 / 6

P(B) = 1 / 2

P(A ∩ B) (X par si suma ≥ 10) = 4 / 36 = 1 / 9

Deoarece P(A) * P(B) = 1 / 12, care este diferit de P(A ∩ B) = 1 / 9, evenimentele NU sunt independente.

Extensie Python: Simulaţi 100.000 de aruncări şi estimaţi probabilităţile condiţionate. Comparaţi rezultatele cu val-
orile exacte.

```python
import numpy as np

# Generam 100.000 de aruncari pentru 2 zaruri
N = 100000
aruncari = np.random.randint(1, 7, size=(N, 2))

X = aruncari[:, 0]
Y = aruncari[:, 1]
suma = X + Y

# Mastile logice (conditiile)
cond1 = (X % 2 == 0)
cond2 = (X % 2 == 0) | (Y % 2 == 0)
cond3 = (suma == 8)

# 1. P(suma >= 10 | X par)
p1_sim = np.mean(suma[cond1] >= 10)

# 2. P(suma >= 10 | X par sau Y par)
p2_sim = np.mean(suma[cond2] >= 10)

# 3. P(X par si Y par | suma == 8)
p3_sim = np.mean((X[cond3] % 2 == 0) & (Y[cond3] % 2 == 0))

# Afisarea rezultatelor in format tabelar simplu
print("1. P(suma >= 10 | X par):          Teoretic = 0.2222 | Simulat =", round(p1_sim, 4))
print("2. P(suma >= 10 | un zar par):     Teoretic = 0.1852 | Simulat =", round(p2_sim, 4))
print("3. P(ambele pare | suma = 8):      Teoretic = 0.6000 | Simulat =", round(p3_sim, 4))
```

Ex. 2 Filtrarea mesajelor spam

Un sistem de e-mail primeşte 30% mesaje spam. Cuvântul „ofertă” apare în 80% dintre mesajele spam şi în 15% dintre
mesajele legitime.

1. Care este probabilitatea ca un mesaj care conţine cuvântul „ofertă” să fie spam?

P(Spam) = 0.30

P(Legitim) = 0.70

P(Oferta | Spam) = 0.80

P(Oferta | Legitim) = 0.15

P(Gratuit | Spam) = 0.60

P(Gratuit | Legitim) = 0.10

Folosim Teorema lui Bayes. Mai intai calculam probabilitatea totala ca un mesaj sa contina cuvantul "oferta":

P(Oferta) = P(Oferta | Spam) * P(Spam) + P(Oferta | Legitim) * P(Legitim)

P(Oferta) = (0.80 * 0.30) + (0.15 * 0.70) = 0.24 + 0.105 = 0.345

Acum aplicam formula:

P(Spam | Oferta) = (P(Oferta | Spam) * P(Spam)) / P(Oferta)

P(Spam | Oferta) = 0.24 / 0.345 ≈ 0.6956 (sau 69.56%)

2. Presupunem că termenul „gratuit” apare în 60% dintre mesajele spam şi în 10% dintre mesajele legitime. Consid-
erând independenţa condiţionată a apariţiei celor două cuvinte, calculaţi probabilitatea ca un mesaj care conţine

ambele cuvinte să fie spam.

Independenta conditionata inseamna ca putem inmulti probabilitatile de aparitie a cuvintelor:

P(Oferta, Gratuit | Spam) = P(Oferta | Spam) * P(Gratuit | Spam) = 0.80 * 0.60 = 0.48

P(Oferta, Gratuit | Legitim) = P(Oferta | Legitim) * P(Gratuit | Legitim) = 0.15 * 0.10 = 0.015

Calculam probabilitatea totala sa apara ambele cuvinte:

P(Oferta, Gratuit) = (0.48 * 0.30) + (0.015 * 0.70) = 0.144 + 0.0105 = 0.1545

Aplicam din nou Bayes:

P(Spam | Oferta, Gratuit) = (0.48 * 0.30) / 0.1545

P(Spam | Oferta, Gratuit) = 0.144 / 0.1545 ≈ 0.9320 (sau 93.20%)

3. Explicaţi de ce ipoteza independenţei condiţionate poate fi nerealistă.

Aceasta ipoteza este nerealista deoarece in limbajul natural cuvintele nu apar independent. Daca un email contine deja cuvantul "oferta", sansele sa apara si "gratuit" sunt mult mai mari decat in mod normal, deoarece ele formeaza expresii comune de marketing (ex: "oferta gratuita"). Independenta presupune ca aparitia unuia nu afecteaza aparitia celuilalt, ceea ce e fals.

4. Propuneţi o regulă de clasificare bazată pe probabilitatea condiţionată.

O regula simpla (cunoscuta ca Naive Bayes) este bazata pe un prag (threshold):

Regula: Un mesaj este clasificat ca SPAM daca P(Spam | cuvinte) > 0.5. Altfel, este considerat Legitim.

Nota: In practica, pentru a nu trimite emailuri importante in spam (rezultat numit "fals pozitiv"), pragul se seteaza de obicei mai sus (ex: > 0.90 sau > 0.95). In cazul de la punctul 2 (0.9320), mesajul ar fi clasificat ca SPAM.

Ex. 3 Paradoxul lui Simpson
O universitate analizează rata de promovare la două discipline, A şi B, pentru studenţii din două programe de studiu:

Program      Disciplina A    Disciplina B
Programul 1    81 din 90     19 din 20
Programul 2 1 din 10 16 din 80


Valorile reprezintă numărul studenţilor promovaţi din totalul celor evaluaţi.

1. Calculaţi rata de promovare la fiecare disciplină, separat pentru fiecare program.

Programul 1:Disciplina A: $81 / 90 = 0.90$ (90%)Disciplina B: $19 / 20 = 0.95$ (95%)

Programul 2:Disciplina A: $1 / 10 = 0.10$ (10%)Disciplina B: $16 / 80 = 0.20$ (20%)

2. Care disciplină are o rată de promovare mai mare în fiecare program?

In ambele programe (1 si 2), Disciplina B are o rata de promovare mai mare decat Disciplina A ($95\% > 90\%$ si $20\% > 10\%$).

3. Calculaţi rata globală de promovare pentru cele două discipline.

Adunam studentii din ambele programe pentru fiecare disciplina:Total Disciplina A: $81 + 1 = 82$ promovati din $90 + 10 = 100$ evaluati.Rata globala A = $82 / 100 = 0.82$ (82%)Total Disciplina B: $19 + 16 = 35$ promovati din $20 + 80 = 100$ evaluati.Rata globala B = $35 / 100 = 0.35$ (35%)

4. Explicaţi de ce ordinea disciplinelor se inversează atunci când agregăm datele.

Desi Disciplina B este mai usoara in fiecare program luat separat, la nivel global Disciplina A pare mult mai usoara. Aceasta inversare (Paradoxul lui Simpson) apare deoarece grupurile sunt dezechilibrate: Disciplina A este sustinuta in mare parte de studentii din Programul 1 (care au rezultate excelente), in timp ce Disciplina B este sustinuta in mare parte de studentii din Programul 2 (care au rezultate foarte slabe).

5. Formulaţi fenomenul în termeni de probabilităţi condiţionate.

Fie P(Promovat|Disciplina, Program). Avem:

P(Promovat|A,P1) < P(Promovat|B,P1)

P(Promovat|A,P2) < P(Promovat|B,P2)

Cu toate acestea, cand eliminam conditionarea de program, relatia se inverseaza:

P(Promovat|A) < P(Promovat|B)

Extensie Python: Construiţi un exemplu propriu de paradox Simpson şi utilizaţi librăria pandas pentru a analiza
datele.

```python
import pandas as pd

# 1. Crearea setului de date
date = {
    'Spital': ['A', 'A', 'B', 'B'],
    'Caz': ['Usor', 'Grav', 'Usor', 'Grav'],
    'Vindecati': [810, 30, 95, 315],
    'Total_Pacienti': [900, 100, 100, 900]
}

df = pd.DataFrame(date)

# 2. Rata de vindecare pe categorii
df['Rata_%'] = (df['Vindecati'] / df['Total_Pacienti']) * 100

print("--- Rata pe categorii ---")
print(df[['Spital', 'Caz', 'Rata_%']])

# 3. Rata de vindecare globala (agregata)
global_df = df.groupby('Spital')[['Vindecati', 'Total_Pacienti']].sum()
global_df['Rata_Globala_%'] = (global_df['Vindecati'] / global_df['Total_Pacienti']) * 100

print("\n--- Rata globala ---")
print(global_df[['Rata_Globala_%']])
```
